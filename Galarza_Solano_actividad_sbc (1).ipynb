## 1. Introducción

### Descripción del problema
El sistema diagnostica problemas que ocurren al enviar documentos electrónicos (facturas, notas crédito, remesas y manifiestos) desde un software empresarial de gestión de transporte hacia entidades externas como la DIAN (Dirección de Impuestos y Aduanas Nacionales) y el RNDC (Registro Nacional de Despachos de Carga). Cuando un documento es rechazado o el envío falla, el sistema identifica la causa del problema a partir de síntomas observables.

### Justificación
Las empresas de transporte y logística están obligadas por ley a reportar sus operaciones a la DIAN y al RNDC. Los errores en estos envíos generan multas, bloqueos operativos y retrasos en la facturación. Automatizar el diagnóstico de estos fallos reduce el tiempo de resolución y permite a usuarios sin conocimiento técnico identificar y reportar el problema correctamente al área de soporte.

### Alcance
El sistema cubre los problemas más frecuentes al enviar facturas, notas crédito, remesas y manifiestos: DIAN o RNDC no disponibles, errores de certificado digital, rango de numeración inválido, problemas de conectividad, datos inválidos en el documento y documentos duplicados. Quedan fuera del alcance los errores internos del software (bugs), configuraciones de red empresarial complejas y la validación del contenido legal de los documentos.

## 3. Formalización en lógica de predicados

### Grupo A — Conectividad y disponibilidad del servicio

**R1a — DIAN caída (Factura):**
$$\forall x \; (Factura(x) \land EnvioDian(x) \land FacturableOk(x) \land \lnot ErrorConexion(x) \land TieneInternet(x) \land CertificadoVigente(x) \land ResolucionVigente(x) \land NumeroEnRango(x) \land DocumentoRechazado(x) \land \lnot TimeoutEnvio(x) \land \lnot FallaSOAP(x) \rightarrow DianCaida(x))$$

*Para toda factura enviada a la DIAN donde toda la información está correcta, hay internet, sin error de conexión ni timeout ni falla SOAP, y el documento es rechazado, se concluye que la DIAN no está disponible.*

**R1b — DIAN caída (Nota crédito):**
$$\forall x \; (NotaCredito(x) \land EnvioDian(x) \land FacturableOk(x) \land \lnot ErrorConexion(x) \land TieneInternet(x) \land CertificadoVigente(x) \land \lnot FacturaAceptadaCliente(x) \land DocumentoRechazado(x) \land \lnot TimeoutEnvio(x) \land \lnot FallaSOAP(x) \rightarrow DianCaida(x))$$

**R2 — RNDC caído (falla SOAP):**
$$\forall x \; ((Manifiesto(x) \lor Remesa(x)) \land EnvioRNDC(x) \land FallaSOAP(x) \rightarrow RndcCaida(x))$$

*Para todo manifiesto o remesa enviado al RNDC que devuelva una falla SOAP, se concluye que el RNDC no está disponible.*

**R3 — Error en conexión con la API:**
$$\forall x \; (Documento(x) \land (EnvioDian(x) \lor EnvioRNDC(x)) \land ErrorConexion(x) \land TieneInternet(x) \land \lnot TimeoutEnvio(x) \rightarrow ErrorAPI(x))$$

**R7 — Problema de internet:**
$$\forall x \; ((EnvioDian(x) \lor EnvioRNDC(x)) \land \lnot TieneInternet(x) \rightarrow ProblemaInternet(x))$$

**R4 — Error de certificado:**

### Grupo C — Numeración DIAN

**R5a — Resolución no vigente:**
$$\forall x \; (Factura(x) \land EnvioDian(x) \land \lnot ResolucionVigente(x) \rightarrow ErrorResolucionNumeracion(x))$$

*Para toda factura enviada a la DIAN cuya resolución de rango de numeración no esté vigente, existe un error de resolución.*

**R5b — Número fuera de rango:**
$$\forall x \; (Factura(x) \land EnvioDian(x) \land \lnot NumeroEnRango(x) \rightarrow ErrorNumeroFueraRango(x))$$

*Para toda factura enviada a la DIAN cuyo número no esté dentro del rango autorizado, existe un error de numeración.*

### Grupo D — Documento duplicado

**R6 — Documento ya procesado:**
$$\forall x \; (Documento(x) \land DocumentoDuplicado(x) \rightarrow DocumentoYaProcesado(x))$$

### Grupo H — Errores en Nota crédito → DIAN

**R13a — Facturable mal diligenciado (Nota crédito):**
$$\forall x \; (NotaCredito(x) \land EnvioDian(x) \land \lnot FacturableOk(x) \rightarrow FacturableMalDiligenciado(x))$$

**R13b — Factura ya aceptada por el cliente:**
$$\forall x \; (NotaCredito(x) \land EnvioDian(x) \land FacturaAceptadaCliente(x) \rightarrow ErrorFacturaYaAceptada(x))$$

*Para toda nota crédito enviada a la DIAN cuya factura asociada ya fue aceptada por el cliente, no se puede emitir la nota crédito.*

### Grupo I — Factura / Nota crédito → RNDC (tipo 12)

**R14a — Documento no es tipo 12 (Factura → RNDC):**
$$\forall x \; (Factura(x) \land EnvioRNDC(x) \land \lnot DocumentoTipo12(x) \rightarrow ErrorDocumentoNoTipo12(x))$$

**R14b — Documento no es tipo 12 (Nota crédito → RNDC):**
$$\forall x \; (NotaCredito(x) \land EnvioRNDC(x) \land \lnot DocumentoTipo12(x) \rightarrow ErrorDocumentoNoTipo12(x))$$

**R14c — Facturable inconsistente (Factura → RNDC):**
$$\forall x \; (Factura(x) \land EnvioRNDC(x) \land DocumentoTipo12(x) \land \lnot FacturableConsistente(x) \rightarrow ErrorFacturableInconsistente(x))$$

**R14d — Facturable inconsistente (Nota crédito → RNDC):**
$$\forall x \; (NotaCredito(x) \land EnvioRNDC(x) \land DocumentoTipo12(x) \land \lnot FacturableConsistente(x) \rightarrow ErrorFacturableInconsistente(x))$$

### Grupo J — Casos de éxito

**R15 — Envío exitoso de manifiesto al RNDC:**
$$\forall x \; (Manifiesto(x) \land EnvioRNDC(x) \land TerceroOk(x) \land FacturableOk(x) \land PropietarioOk(x) \land ClienteOk(x) \land DestinatarioOk(x) \land RemitenteOk(x) \land VehiculoOk(x) \land TieneInternet(x) \land \lnot FallaSOAP(x) \rightarrow EnvioExitosoRNDC(x))$$

*El RNDC devuelve número de autorización cuando todos los datos de las entidades del manifiesto son correctos y el servicio está disponible.*

**R16a — CUFE generado (Factura):**
$$\forall x \; (Factura(x) \land EnvioDian(x) \land CertificadoVigente(x) \land ResolucionVigente(x) \land NumeroEnRango(x) \land FacturableOk(x) \rightarrow CufeGenerado(x))$$

**R16b — CUFE generado (Nota crédito):**
$$\forall x \; (NotaCredito(x) \land EnvioDian(x) \land CertificadoVigente(x) \land FacturableOk(x) \land \lnot FacturaAceptadaCliente(x) \rightarrow CufeGenerado(x))$$

**R17a — XML tipo 12 (Factura):**
$$\forall x \; (Factura(x) \land CufeGenerado(x) \land DocumentoTipo12(x) \rightarrow TipoXML12(x))$$

**R17b — XML tipo 10 (Factura):**
$$\forall x \; (Factura(x) \land CufeGenerado(x) \land \lnot DocumentoTipo12(x) \rightarrow TipoXML10(x))$$

**R17c — XML tipo 12 (Nota crédito):**
$$\forall x \; (NotaCredito(x) \land CufeGenerado(x) \land DocumentoTipo12(x) \rightarrow TipoXML12(x))$$

**R17d — XML tipo 10 (Nota crédito):**
$$\forall x \; (NotaCredito(x) \land CufeGenerado(x) \land \lnot DocumentoTipo12(x) \rightarrow TipoXML10(x))$$

## 7. Motor de inferencia

````python
PREDICADOS_SINTOMA = {
    # Tipo de documento
    "Documento", "Factura", "NotaCredito", "Remesa", "Manifiesto",
    # Destino del envío
    "EnvioDian", "EnvioRNDC",
    # Validez de entidades (reemplazan al anterior InformacionOk)
    "TerceroOk", "PropietarioOk", "ClienteOk",
    "DestinatarioOk", "RemitenteOk", "VehiculoOk", "FacturableOk",
    # Condiciones DIAN
    "CertificadoVigente", "ResolucionVigente", "NumeroEnRango",
    # Estado del envío
    "ErrorConexion", "TimeoutEnvio", "FallaSOAP",
    "DocumentoDuplicado", "DocumentoRechazado",
    # Conectividad
    "TieneInternet",
    # Condiciones RNDC — remesa
    "CoordenadasCoherentes", "DestinoExisteRNDC",
    # Condiciones nota crédito
    "FacturaAceptadaCliente",
    # Condiciones Factura/NC → RNDC
    "DocumentoTipo12", "FacturableConsistente",
}

TIPOS_DOCUMENTO = {"Factura", "NotaCredito", "Remesa", "Manifiesto"}

CONFLICTOS_DIAGNOSTICO = [
    frozenset({"DianCaida",        "DocumentoYaProcesado"}),
    frozenset({"RndcCaida",        "DocumentoYaProcesado"}),
    frozenset({"ProblemaInternet", "DianCaida"}),
    frozenset({"ProblemaInternet", "RndcCaida"}),
    frozenset({"ProblemaInternet", "ErrorAPI"}),
    frozenset({"EnvioExitosoRNDC", "RndcCaida"}),
    frozenset({"EnvioExitosoRNDC", "TerceroMalDiligenciado"}),
    frozenset({"EnvioExitosoRNDC", "PropietarioMalDiligenciado"}),
    frozenset({"EnvioExitosoRNDC", "ClienteMalDiligenciado"}),
    frozenset({"EnvioExitosoRNDC", "VehiculoMalDiligenciado"}),
    frozenset({"EnvioExitosoRNDC", "FacturableMalDiligenciado"}),
    frozenset({"CufeGenerado",     "DianCaida"}),
    frozenset({"CufeGenerado",     "ErrorCertificado"}),
    frozenset({"TipoXML10",        "TipoXML12"}),
]


def validar_hechos(hechos):
    """Lanza ValueError si hay predicados contradictorios o tipos de documento mutuamente excluyentes."""
    vistos = {}
    for pred, valor in hechos:
        if pred in vistos and vistos[pred] != valor:
            raise ValueError(
                f"Hecho contradictorio: '{pred}' aparece como True y False al mismo tiempo."
            )
        vistos[pred] = valor

    tipos_activos = [t for t in TIPOS_DOCUMENTO if vistos.get(t) is True]
    if len(tipos_activos) > 1:
        raise ValueError(
            f"Tipos de documento mutuamente excluyentes: {', '.join(tipos_activos)}. "
            "Un documento solo puede ser de un tipo a la vez."
        )


def aplicar_mundo_cerrado(hechos):
    """Agrega (pred, False) para cada predicado de síntoma no provisto explícitamente."""
    predicados_presentes = {h[0] for h in hechos}
    return list(hechos) + [
        (pred, False)
        for pred in PREDICADOS_SINTOMA
        if pred not in predicados_presentes
    ]


def detectar_conflictos_diagnostico(diagnosticos):
    """Retorna grupos de diagnósticos mutuamente excluyentes que coexisten."""
    nombres = {d[0] for d in diagnosticos if d[1]}
    return [g for g in CONFLICTOS_DIAGNOSTICO if g.issubset(nombres)]
````

````python
def encadenamiento_adelante(reglas, hechos):
    """Aplica encadenamiento hacia adelante con validación, mundo cerrado y certeza."""
    # Eliminar hechos duplicados preservando orden
    hechos_unicos = list(dict.fromkeys(hechos))

    # Rechazar hechos contradictorios o tipos de documento inválidos antes de inferir
    validar_hechos(hechos_unicos)

    # Supuesto de mundo cerrado — asumir False para síntomas no provistos
    nuevos_hechos = aplicar_mundo_cerrado(hechos_unicos)

    reglas_aplicadas = []
    certezas_derivadas = {}
    cambio = True

    while cambio:
        cambio = False
        for regla in reglas:
            if regla["nombre"] in reglas_aplicadas:
                continue
            cumple = all(c in nuevos_hechos for c in regla["condiciones"])
            if cumple:
                # Registrar la regla aunque la conclusión ya haya sido derivada por otra
                reglas_aplicadas.append(regla["nombre"])
                if regla["conclusion"] not in nuevos_hechos:
                    nuevos_hechos.append(regla["conclusion"])
                    certezas_derivadas[regla["conclusion"]] = regla["certeza"]
                    cambio = True
                else:
                    # Combinar certezas cuando múltiples reglas derivan la misma conclusión.
                    # Usando 0 como base: sin evidencia previa, la certeza parte desde 0
                    # y la fórmula P(A∨B)=1-(1-P(A))*(1-P(B)) devuelve la certeza de la regla.
                    c_existente = certezas_derivadas.get(regla["conclusion"], 0)
                    certezas_derivadas[regla["conclusion"]] = round(
                        1 - (1 - c_existente) * (1 - regla["certeza"]), 4
                    )

    return nuevos_hechos, reglas_aplicadas, certezas_derivadas
````

````python
DESCRIPCIONES_REGLAS = {
    # Grupo A — Conectividad
    "R1a":  "Factura+DIAN+FacturableOk+SinError+Internet+CertVig+ResolVig+NumRango+Rechazado+SinTimeout+SinSOAP → DIAN caída",
    "R1b":  "NC+DIAN+FacturableOk+SinError+Internet+CertVig+SinFactAceptada+Rechazado+SinTimeout+SinSOAP    → DIAN caída",
    "R2a":  "Manifiesto+RNDC+FallaSOAP                                                                      → RNDC caído",
    "R2b":  "Remesa+RNDC+FallaSOAP                                                                          → RNDC caído",
    "R3a":  "Doc+DIAN+ErrorConexion+Internet+SinTimeout                                                     → Error API",
    "R3b":  "Doc+RNDC+ErrorConexion+Internet+SinTimeout                                                     → Error API",
    "R7a":  "EnvioDIAN+SinInternet                                                                          → Problema internet",
    "R7b":  "EnvioRNDC+SinInternet                                                                          → Problema internet",
    # Grupo B — Certificado
    "R4a":  "Factura+DIAN+CertificadoNoVigente                                                              → Error certificado",
    "R4b":  "NC+DIAN+CertificadoNoVigente                                                                   → Error certificado",
    # Grupo C — Numeración
    "R5a":  "Factura+DIAN+ResolucionNoVigente                                                               → Error resolución numeración",
    "R5b":  "Factura+DIAN+NumeroFueraRango                                                                  → Error número fuera de rango",
    # Grupo D — Duplicado
    "R6":   "Doc+Duplicado                                                                                   → Documento ya procesado",
    # Grupo E — Manifiesto entidades
    "R10a": "Manifiesto+RNDC+TerceroInválido                                                                → Tercero mal diligenciado",
    "R10b": "Manifiesto+RNDC+PropietarioInválido                                                            → Propietario mal diligenciado",
    "R10c": "Manifiesto+RNDC+ClienteInválido                                                                → Cliente mal diligenciado",
    "R10d": "Manifiesto+RNDC+VehiculoInválido                                                               → Vehículo mal diligenciado",
    "R10e": "Manifiesto+RNDC+FacturableInválido                                                             → Facturable mal diligenciado",
    # Grupo F — Remesa errores
    "R11a": "Remesa+RNDC+TerceroInválido                                                                    → Tercero mal diligenciado",
    "R11b": "Remesa+RNDC+DestinatarioInválido                                                               → Destinatario mal diligenciado",
    "R11c": "Remesa+RNDC+DestinoNoExiste                                                                    → Destino no existe en RNDC",
    "R11d": "Remesa+RNDC+CoordenadasIncoherentes                                                            → Error coordenadas remesa",
    # Grupo G — Factura DIAN datos
    "R12":  "Factura+DIAN+FacturableInválido                                                                → Facturable mal diligenciado",
    # Grupo H — NC DIAN errores
    "R13a": "NC+DIAN+FacturableInválido                                                                     → Facturable mal diligenciado",
    "R13b": "NC+DIAN+FacturaYaAceptada                                                                      → Error factura ya aceptada",
    # Grupo I — Factura/NC → RNDC tipo 12
    "R14a": "Factura+RNDC+NoEsTipo12                                                                        → Error documento no tipo 12",
    "R14b": "NC+RNDC+NoEsTipo12                                                                             → Error documento no tipo 12",
    "R14c": "Factura+RNDC+Tipo12+FacturableInconsistente                                                   → Error facturable inconsistente",
    "R14d": "NC+RNDC+Tipo12+FacturableInconsistente                                                        → Error facturable inconsistente",
    # Grupo J — Éxito
    "R15":  "Manifiesto+RNDC+TodasEntidadesOk+Internet+SinSOAP                                             → Envío exitoso RNDC",
    "R16a": "Factura+DIAN+CertVig+ResolVig+NumRango+FacturableOk                                           → CUFE generado",
    "R16b": "NC+DIAN+CertVig+FacturableOk+SinFactAceptada                                                  → CUFE generado",
    "R17a": "Factura+CufeGenerado+Tipo12                                                                    → XML tipo 12",
    "R17b": "Factura+CufeGenerado+NoTipo12                                                                  → XML tipo 10",
    "R17c": "NC+CufeGenerado+Tipo12                                                                         → XML tipo 12",
    "R17d": "NC+CufeGenerado+NoTipo12                                                                       → XML tipo 10",
}

# Predicados que son diagnósticos (no síntomas)
PREDICADOS_DIAGNOSTICO = {
    "DianCaida", "RndcCaida", "ErrorAPI", "ErrorCertificado",
    "ErrorResolucionNumeracion", "ErrorNumeroFueraRango", "DocumentoYaProcesado",
    "ProblemaInternet", "TerceroMalDiligenciado", "PropietarioMalDiligenciado",
    "ClienteMalDiligenciado", "DestinatarioMalDiligenciado", "VehiculoMalDiligenciado",
    "FacturableMalDiligenciado", "ErrorCoordenadasRemesa", "DestinoNoExisteRNDC",
    "ErrorFacturaYaAceptada", "ErrorDocumentoNoTipo12", "ErrorFacturableInconsistente",
    "EnvioExitosoRNDC", "CufeGenerado", "TipoXML12", "TipoXML10",
}


def explicar_razonamiento(hechos_iniciales, hechos_finales, reglas_aplicadas, certezas_derivadas):
    """Imprime el resumen del diagnóstico: hechos ingresados, reglas y conclusiones."""
    print("=" * 60)
    print("DIAGNÓSTICO DEL SISTEMA")
    print("=" * 60)

    print("\nHechos ingresados:")
    for h in hechos_iniciales:
        estado = "SÍ" if h[1] else "NO"
        print(f"  {estado} {h[0]}")

    print("\nReglas aplicadas:")
    if reglas_aplicadas:
        for r in reglas_aplicadas:
            print(f"  [{r}] {DESCRIPCIONES_REGLAS.get(r, '')}")
    else:
        print("  Ninguna regla fue disparada.")

    diagnosticos = [
        h for h in hechos_finales
        if h not in hechos_iniciales and h[1] and h[0] in PREDICADOS_DIAGNOSTICO
    ]
    errores    = [d for d in diagnosticos if d[0] not in {"EnvioExitosoRNDC", "CufeGenerado", "TipoXML12", "TipoXML10"}]
    exitos     = [d for d in diagnosticos if d[0] in {"EnvioExitosoRNDC", "CufeGenerado", "TipoXML12", "TipoXML10"}]

    if exitos:
        print("\nResultados de éxito:")
        for d in exitos:
            certeza = certezas_derivadas.get(d, 1.0)
            print(f"  ✓ {d[0]}  (certeza: {certeza:.0%})")

    print("\nDiagnósticos de error:")
    if errores:
        for d in errores:
            certeza = certezas_derivadas.get(d, 1.0)
            print(f"  >>> {d[0]}  (certeza: {certeza:.0%})")
        conflictos = detectar_conflictos_diagnostico(diagnosticos)
        if conflictos:
            print("\n  ADVERTENCIA — diagnósticos mutuamente excluyentes detectados:")
            for grupo in conflictos:
                print(f"    {' vs '.join(grupo)}")
    else:
        if not exitos:
            print("  No se detectaron problemas con los síntomas proporcionados.")
        else:
            print("  Ninguno.")
    print("=" * 60)


def porque(conclusion_buscada, reglas, reglas_aplicadas):
    """Muestra todas las reglas aplicadas que derivaron la conclusión indicada."""
    print(f"\n¿Por qué se concluyó '{conclusion_buscada}'?")
    encontrada = False
    for regla in reglas:
        if regla["nombre"] in reglas_aplicadas and regla["conclusion"][0] == conclusion_buscada:
            print(f"  Regla {regla['nombre']}: {DESCRIPCIONES_REGLAS.get(regla['nombre'], '')}")
            print("  Condiciones que se cumplieron:")
            for cond in regla["condiciones"]:
                estado = "SÍ" if cond[1] else "NO"
                print(f"    - {estado} {cond[0]}")
            print(f"  Factor de certeza: {regla['certeza']}")
            encontrada = True
    if not encontrada:
        print(f"  '{conclusion_buscada}' no fue derivada por ninguna regla aplicada.")
````

## 11. Análisis y conclusiones

### Limitaciones del sistema

- **Sensibilidad al supuesto de mundo cerrado (CWA):** el motor asume `False` para todo síntoma no provisto explícitamente. Esto exige que el usuario conozca y declare todos los síntomas relevantes; omitir uno puede generar falsos positivos (por ejemplo, no declarar `FacturableOk=True` dispararía R12 erróneamente al evaluar un caso de internet sin conexión).
- **Condiciones binarias:** no considera estados intermedios como «certificado próximo a vencer», «internet intermitente» o «RNDC con alta latencia».
- **Sin memoria histórica:** no distingue si un error de tercero ocurre por primera vez o si es recurrente en un cliente específico, ni si el RNDC ha estado inestable durante el día.
- **Certezas fijas:** los factores de certeza son asignados manualmente y no se ajustan con el historial real de fallos del sistema.
- **Encadenamiento lineal:** el motor no maneja negación de conclusiones derivadas; si `CufeGenerado` se infiere y luego se detecta un error lógico, ambos coexistirían hasta que el usuario lo corrija.

### Posibles mejoras

- Integrar el sistema con los logs del software de transporte para capturar los hechos automáticamente (respuestas SOAP del RNDC, respuestas CUFE de la DIAN) sin intervención del usuario.
- Implementar encadenamiento hacia atrás para hacer preguntas dirigidas al operario (¿el certificado está vigente? ¿el número de factura está en el rango?) en lugar de pedirle todos los síntomas a la vez.
- Agregar certezas dinámicas basadas en el historial de errores de cada cliente, permitiendo priorizar diagnósticos más probables según el contexto.
- Extender el modelo para cubrir errores de coordenadas específicos por ciudad, basados en una base de datos geográfica del RNDC.
- Distinguir entre errores transitorios (RNDC caído) y errores permanentes (facturable con NIT incorrecto) para dar recomendaciones de acción diferenciadas.

### Reflexión sobre el proceso

El dominio de integración con la DIAN y el RNDC resultó muy adecuado para un SBC porque las causas de error son finitas, bien documentadas y siguen flujos predecibles. La expansión del sistema para incorporar predicados granulares por entidad (tercero, propietario, cliente, vehículo, facturable) refleja directamente el conocimiento del equipo de soporte al atender tickets: la primera pregunta siempre es «¿cuál entidad tiene el dato incorrecto?». El mayor reto técnico fue mantener la coherencia entre la formalización lógica, las reglas de producción y el código cuando el número de reglas creció de 9 a 35, y garantizar que el supuesto de mundo cerrado no generara falsos positivos al introducir predicados de éxito que se encadenan con síntomas.
